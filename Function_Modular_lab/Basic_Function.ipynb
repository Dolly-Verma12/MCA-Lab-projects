```python
def factorial(n):

    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)

print("Factorial:", factorial(5))
```

```python
def fibonacci(n):

    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(8):
    print(fibonacci(i), end=" ")
```

```python
def recursive_sum(n):

    if n == 0:
        return 0

    return n + recursive_sum(n - 1)

print("Sum:", recursive_sum(5))
```

```python

def calculate_sum(*numbers):

    total = 0

    for number in numbers:
        total += number

    return total

print(calculate_sum(10, 20))
print(calculate_sum(10, 20, 30, 40))
```

```python
def find_maximum(*numbers):

    if not numbers:
        return None

    maximum = numbers[0]

    for number in numbers[1:]:
        if number > maximum:
            maximum = number

    return maximum

print("Maximum:", find_maximum(10, 50, 20, 80, 30))
```

```python

def display_employee(**details):

    for key, value in details.items():
        print(key, ":", value)

display_employee(
    name="Ravi",
    department="Development",
    experience=2,
    location="Hyderabad"
)
```

```python
def employee_report(*skills, **details):

    print("Employee Details:")

    for key, value in details.items():
        print(key, ":", value)

    print("\nSkills:")

    for skill in skills:
        print("-", skill)

employee_report(
    "Python",
    "SQL",
    "Git",
    name="Ravi",
    department="Development"
)
```
